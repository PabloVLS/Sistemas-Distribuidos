```python
import requests
response = requests.get("https://iftm.edu.br/noticias/tag/?tag=DADOS%20ABERTOS")
if response.status_code != 200:
  print("Error fetching page")
  exit()
else:
  content = response.content
```

```python
print(content)
```

```python
from bs4 import BeautifulSoup
sout = BeautifulSoup(response.content, 'html.parser')
```

```python
sout.find_all("p")
```

```python
for i in range(len(sout.find_all("div"))-1):
  print(sout.find_all("div")[i].getText())
  print()
```
